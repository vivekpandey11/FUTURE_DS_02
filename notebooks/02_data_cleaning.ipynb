```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

print("Original Shape:", df.shape)
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing.sort_values(
    by="Missing Percentage",
    ascending=False
)
```

```python
print("Duplicate Rows:", df.duplicated().sum())

print("Duplicate Customer IDs:",
      df["customerID"].duplicated().sum())
```

```python
df.dtypes
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        print(col, ":", (df[col].astype(str).str.strip() == "").sum())
```

```python
print(df["Churn"].value_counts())

print("\nChurn Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
# Remove extra spaces
df["TotalCharges"] = df["TotalCharges"].astype(str).str.strip()

# Convert column into numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
)

print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())
```

```python
missing_rows = df[df["TotalCharges"].isnull()]

missing_rows[["customerID", "tenure", "MonthlyCharges", "TotalCharges"]]
```

```python
df.loc[
    (df["TotalCharges"].isnull()) & (df["tenure"] == 0),
    "TotalCharges"
] = 0

print("Remaining Missing Values:")
print(df.isnull().sum().sum())
```

```python
df["SeniorCitizen"] = df["SeniorCitizen"].astype(int)

print(df.dtypes)
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nChurn Distribution:")
print(df["Churn"].value_counts())
```

```python
df.to_csv(
    "../data/processed/cleaned_customer_churn.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```
